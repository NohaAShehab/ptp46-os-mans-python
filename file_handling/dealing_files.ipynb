# read data

```python
try:
    file_obj = open("users.txt", "r")
except Exception as e:
    print(e)
else:
    print(file_obj)
    data = file_obj.read() # read file content from byte 0 -==> into one string
    print(data)
    file_obj.close()
```

```python
try:
    file_obj = open("users.txt", "r")
except Exception as e:
    print(e)
else:
    print(file_obj)
    data = file_obj.read() # read file content from byte 0 -==> into one string
    print(data)
    file_obj.seek(0)  # reset pointer to the beginning of file
    dat2 = file_obj.readlines() #read file content to a list ?
    print(dat2)
    file_obj.close()
```

```python
try:
    with open("users.txt", "r") as f:
        print(f.read())

except Exception as e:
    print(e)
```

# open for write

```python
# open for read
"""
    if you are opening file with w mode, if file not exists,, python will try to create
    it  .. .
    if file exists  ---> python open file for writing starting from the beginning of the
    file --> remove old content
"""
try:
    file_obj = open("students.txt", "w")
except Exception as e:
    print(e)
else:
    print(file_obj)
    file_obj.close()
```

```python
try:
    file_obj = open("students.txt", "w")
except Exception as e:
    print(e)
else:
    print(file_obj)
    file_obj.write("Hello from python\n")
    file_obj.write("Hello from python ")
    file_obj.close()
```

```python
try:
    file_obj = open("students.txt", "w")
except Exception as e:
    print(e)
else:
    print(file_obj)
    file_obj.write("Hello from python\n")
    file_obj.write("Hello from python ")
    file_obj.writelines(["Hello\n", "world\n", "Python"])
    file_obj.close()
```

# append

```python
"""
    if you are opening file with a mode, if file not exists,, python will try to create
    it  .. .
    if file exists  ---> python open file for appending starting from the end of the
    file --> keep  old content
"""
try:
    file_obj = open("students.txt", "a")
except Exception as e:
    print(e)
else:
    print(file_obj)
    file_obj.write("Hello from python\n")
    file_obj.write("Hello from python ")
    file_obj.writelines(["Hello\n", "world\n", "Python\n"])
    file_obj.close()
```
