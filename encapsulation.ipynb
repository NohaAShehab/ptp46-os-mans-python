```python

class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        self.set_salary(salary)

    def display(self):
        print(f"Employee Name: {self.name}, Employee Email: {self.email}, Salary: {self.salary}")

    @property
    def salary(self): # get value of __salary
        return self.__salary

    # set value of __salary
    @salary.setter
    def salary(self, salary):
        if isinstance(salary, float)  or isinstance(salary, int)  and salary > 0 :
            self.__salary = salary
        else:
            raise TypeError("Salary should be a float or int and > 0")


    def set_salary(self, salary):
        if isinstance(salary, float)  or isinstance(salary, int)  and salary > 0 :
            self.__salary = salary
        else:
            raise TypeError("Salary should be a float or int and > 0")



emp = Employee("Noha", "noha@gmail.com", 5454)
emp.set_salary(2423) # set value of property _Employee__salary

print(emp.salary)  # ... get value
print(emp.__dict__)
emp.salary = 55000  # call salary.setter
```

```python

class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        # self.set_salary(salary)
        self.salary = salary # call setter


    def display(self):
        print(f"Employee Name: {self.name}, Employee Email: {self.email}, Salary: {self.salary}")

    @property
    def salary(self): # get value of __salary
        return self.__salary

    # set value of __salary
    @salary.setter
    def salary(self, salary):
        if isinstance(salary, float)  or isinstance(salary, int)  and salary > 0 :
            self.__salary = salary
        else:
            raise TypeError("Salary should be a float or int and > 0")

    @salary.deleter
    def salary(self):
        del self.__salary






emp = Employee("Noha", "noha@gmail.com", 5454)
print(emp.salary)  # ... get value
print(emp.__dict__)
emp.salary = 55000  # call salary.setter
del emp.salary
print(emp.__dict__)
```
