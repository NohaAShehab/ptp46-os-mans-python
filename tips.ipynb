```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self.email = email
        self.salary = salary

    def __str__(self):
        return f"{self.name}"

    def __repr__(self):
        return f"Employee (name= {self.name}, email= {self.email}, salary= {self.salary})"

    def __len__(self):
        # must return with int
        return len(self.__dict__)
emp = Employee("ee", "ee", 222)
print(emp) # check __str__ , __repr__ , else print object address
emp.city = 'abc'
print(len(emp))
```
