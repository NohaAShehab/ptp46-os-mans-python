## 📋 List Example - Unlabeled Data
> 🎯 **Creating a simple list** with mixed data types (strings and numbers) to demonstrate unlabeled data structure

```python
info = ["noha", "iti", "open source", 33, "Mansoura"]
print(info) # unlabeled
```

## 📚 Dictionary Basics
> 🔑 **Creating empty dictionaries** using two different methods: literal `{}` and constructor `dict()`

```python
# datatype --> keyword: value ===> dictionary , comma seperated key:value pair
d = {}
myd = dict()
```

## 🔑 Important Dictionary Properties
> ⚠️ **No key duplication:** Each key must be unique - duplicate keys keep only the last value
> 
> 📊 **Ordered since Python 3.6+:** Dictionaries maintain insertion order!

## 🎭 Key Duplication Demo
> 🔄 **Testing duplicate keys:** When the same key appears multiple times, only the last value is kept!

```python
my_info = {
    "name":"noha",
    "track":"cloud",
    "age":33,
    "name": "Noha Shehab"
}

print(my_info)
```

## 🔍 Accessing Values by Keys
> 📌 **Square bracket notation:** Use `dict[key]` to access values
>
> 💡 **Mutable:** Dictionaries can be modified after creation

```python
"""dict is mutable datatype you access elements using keys """
print(my_info["name"])
print(my_info["track"])
print(my_info["age"])
```

## ✏️ Updating Dictionary Values
> 🔧 **Modifying existing keys:** Change the value associated with an existing key

```python
my_info['track']= 'Cloud Development'  # update
print(my_info)
```

## ➕ Adding New Key-Value Pairs
> ✨ **Dynamic addition:** If a key doesn't exist, assignment will create a new key-value pair!

```python
my_info["city"] = "Mansoura" # if key doesn't exist will add the key, value pair
print(my_info)
```

## 🔍 Accessing Dictionary Values
> 🎯 **Retrieving values by keys:** Dictionaries are mutable, use square brackets with keys to access values

```python
print(len(my_info))
```

```python
print("Mansoura" in my_info) # check if element exists in key
```

```python
print("name" in my_info)
```

```python
# get keys, values, pairs
```

```python
for k  in my_info:
    print(f"key = {k}")
```

```python
for k  in my_info:
    print(f"key = {k}, value = {my_info[k]}")
```

## 📏 Dictionary Length
> 🔢 **Counting items:** Use `len()` to get the number of key-value pairs in the dictionary

```python
info_keys = my_info.keys()
print(info_keys)  # dict_keys
```

```python
print(list(info_keys))
```

## 🔎 Membership Test (Keys Only)
> ⚡ **Checking existence:** The `in` operator checks if a value exists as a KEY, not as a value!

```python
info_values = my_info.values()
print(info_values)
print(list(info_values))
```

```python
info_items = my_info.items()
print(info_items)
info_items_list = list(info_items)
print(info_items_list)
```

## ✅ Testing Key Existence
> 🎯 **Confirming key presence:** Checking if "name" exists as a key in the dictionary

```python
for k , v in my_info.items():
    print(f"{k} = {v}")
```

```python
# pop value from dict ?
popped_value = my_info.pop("name")
print(popped_value)
print(my_info)
```
