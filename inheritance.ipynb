```python
""" single inheritance """
class Person:
    def __init__(self):
        print("--- Person object is created ---")
        self.name = "person"

    def display(self):
        print(f"--- name: {self.name} ---")

class Engineer(Person):
    #  override  the constructor of object
    def __init__(self):
        # u need to call parent constructor
        super().__init__() # call __init__ of the parent
        print("--- Engineer object is created ---")
        self.salary = 1000

e = Engineer()
print(e.__dict__)
e.display()
```

```python
""" single inheritance """
class Person:
    def __init__(self):
        print("--- Person object is created ---")
        self.name = "person"

    def display(self):
        print(f"--- name: {self.name} ---")

class Engineer(Person):
    #  override  the constructor of object
    def __init__(self):
        # u need to call parent constructor
        super().__init__() # call __init__ of the parent
        print("--- Engineer object is created ---")
        self.salary = 1000

    # overriding
    def display(self):
        print(f"---- salary : {self.salary} ---")


e = Engineer()
# print(e.__dict__)
e.display()
```

```python
""" single inheritance """
class Person:
    def __init__(self):
        print("--- Person object is created ---")
        self.name = "person"

    def display(self):
        print(f"--- name: {self.name} ---")

class Engineer(Person):
    #  override  the constructor of object
    def __init__(self):
        # u need to call parent constructor
        super().__init__() # call __init__ of the parent
        print("--- Engineer object is created ---")
        self.salary = 1000

    # overriding
    def display(self):
        super().display()
        print(f"---- salary : {self.salary} ---")


e = Engineer()
# print(e.__dict__)
e.display()
```
