```python
def sum_num(num1 :int, num2 :int) -> int | None:
    # you must validate the data on your own >>
    if isinstance(num1, int) and isinstance(num2, int):
        print(f"num1 = {num1} ,num2 {num2}")
        res = num1 + num2
        print(res)
        return res
    print("--- num1, num2 must be integers ")


r = sum_num(3,5)
print(r)

r2 = sum_num("24", "iti")
print(r2)

r3 = sum_num("323",12)
print(r3)
```
