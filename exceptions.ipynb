```python
def div_nums(num1, num2):
    if num2 ==0 :
        raise ZeroDivisionError("Division by zero Error ")
    if isinstance(num1, int) and isinstance(num2, int):
        print(f"num1 = {num1}, num2 = {num2} ")
        res = num1 / num2
        return res
    else:
        raise TypeError("num1, num2 must be integers ")
```
