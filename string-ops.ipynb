## 🔄 String Replacement

### Method 1: Manual Replacement (Using Loop)

Here's a manual approach to replace characters using a loop:

```python
# Iterate through each character
# Replace matching characters
# Build new string
```

**Note:** This is less efficient than using built-in methods.


````python
# string is iterable
for char in 'noha':
    print(char)
````

````python
mystr = "iti"
for char in mystr:
    print(char, mystr.index(char))
````

````python
mystr = "iti"
counter  = 0
for char in mystr:
    print(char, counter)
    counter +=1
````

````python
message = "The lecture is interesting"
newstr = ''
for c in message:
    if c == ' ':
        newstr += "_"
    else:
        newstr  += c

print(newstr)
````

### Method 2: Using `.replace()` Method ⚡

The `.replace()` method is **much simpler and more efficient and return with newstring**:

```python
text.replace(old, new)  # Replace all occurrences
text.replace(old, new, count)  # Replace first 'count' occurrences
```

**Advantages:**
- ✅ One line of code
- ✅ More efficient
- ✅ Built-in and optimized


---

> **💡 Tip:** You can limit the number of replacements!

Specify a third parameter `count` to replace only the first N occurrences:

```python
text.replace("o", "@", 2)  # Replace only first 2 occurrences
```


````python
message = "The lecture is interesting"
print(message.replace(" ", '_'))
````

````python
print(message.replace(" ", '_', 2))
````

> **⚠️ Common Mistake:** String Concatenation vs Addition

When you use `+` on two strings from `input()`, they are **concatenated**, not added mathematically!

```python
num1 = input("Enter num1: ")  # "3" (string)
num2 = input("Enter num2: ")  # "3" (string)
result = num1 + num2  # "33" (concatenated, not 6!)
```


---

To perform mathematical operations, you must convert the strings to numbers:

```python
num1 = int(input("Enter num1: "))
num2 = int(input("Enter num2: "))
result = num1 + num2  # Now it's addition: 3 + 3 = 6
```

## ✅ String Validation Methods

Python provides several methods to check the content of strings:

| Method | Description | Example |
|--------|-------------|---------|
| `isdigit()` | Returns `True` if all characters are digits | `"123".isdigit()` → `True` |

````python
num1 = input("Enter num1: ")
num2 = input("Enter num2: ")
if num1.isdigit() and num2.isdigit():
    result = num1 + num2  # Now it's addition: 3 + 3 = 6
    print(result)
else:
    print("Num1 and num2 must be integers")
````

## ✂️ Stripping Whitespace

The `rstrip()` method removes whitespace (or specified characters) from the **right end** of a string.

**Related Methods:**
- `rstrip()` - Remove from right
- `lstrip()` - Remove from left  
- `strip()` - Remove from both ends


````python
username = input("Please enter your username: ")
print(username)
````

````python
print(len(username))
````

````python
# strip ??
stripped_username = username.strip()
print(stripped_username, len(stripped_username))
````

````python
stripped_username = username.rstrip() # remove extra-spaces from right
print(stripped_username, len(stripped_username))
````

---

> **💡 Custom Characters:** You can specify which characters to strip!

```python
text = "@     welcome     @"
text.rstrip("@")  # Removes only '@' from the right
```

**Note:** Only removes characters from the **end** of the string.


````python
text = "@     welcome     @"
print(text.rstrip("@"))
````

---

> **⚠️ Important:** `rstrip()` only removes characters from the **end**!

If the specified characters are not at the end, nothing is removed:

```python
text = "  welcome  $  "
text.rstrip("@$")  # Nothing removed (no @ or $ at the end)
```
