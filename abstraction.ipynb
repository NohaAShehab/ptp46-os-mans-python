```python
# abstract class ? 

"""
    abstract class 
    inherit from class ABC , abstract base class 
    --> abstract class --> has at least one abstract method 
"""

from abc import ABC, abstractmethod
class Person(ABC):
    @abstractmethod
    def speak(self):pass 
    
p = Person()
```

```python
class Teacher(ABC):
    def hello(self):
        print("Hello")
        
t = Teacher()
```

```python
class Engineer:
    @abstractmethod
    def hi(self):pass 
    
eng = Engineer()
```
