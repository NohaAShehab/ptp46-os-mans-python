```python
# access class variable using class ---> modify or read -->
# Employee.location ='abc'

# instances of the class can see the class variables

class Employee:
    count = 0  # class variable ?
    location = 'Egypt'

    def __init__(self, id, name, salary):
        self.id = id
        self.name = name
        self.salary = salary
        self.__class__.count +=1


    def display_emp(self):
        print(f"Employee : id= {self.id}, name  = {self.name}, salary = {self.salary}")

    # class method ?? accessed without instance
    # higher order function >> ? accept function --> return new function with different behaviour
    @classmethod  # tell the class that this function depends on class , access it using class name no need to have an object
    def get_number_of_employees(cls): # Usually first parameter of such methods is named 'cls'
        print(cls) # referes to the class itself
        # return Employee.count
        return cls.count


print(Employee.get_number_of_employees())

emp = Employee(1, "test", 10000)
print(Employee.get_number_of_employees())

print(emp.get_number_of_employees()) # this not clean code
```

# Static methods

```python
class Employee:
    count = 0
    location = 'Egypt'

    def __init__(self, id, name, salary):
        self.id = id
        self.name = name
        self.salary = salary
        self.__class__.count +=1


    def display_emp(self):
        print(f"Employee : id= {self.id}, name  = {self.name}, salary = {self.salary}")


    """use class methods in actions only class can do ? """
    # object factory ?
    @classmethod
    def create_default_employee(cls):
        return cls(1, "test", 10000)

emp1 = Employee(1, "Noha", 10000)

# helper method
def cal_net_salary(salary):
    return salary * .8

print(cal_net_salary(emp1.salary))
```

```python
class Employee:
    count = 0
    location = 'Egypt'

    def __init__(self, id, name, salary):
        self.id = id
        self.name = name
        self.salary = salary
        self.__class__.count +=1

    def display_emp(self):
        print(f"Employee : id= {self.id}, name  = {self.name}, salary = {self.salary}")

    """use class methods in actions only class can do ? """
    # object factory ?
    @classmethod
    def create_default_employee(cls):
        return cls(1, "test", 10000)

    # this a helper method, doesn't depend on instance or class ?
    @staticmethod
    def cal_net_salary(salary):
        # print(salary)
        return salary * .8

emp1 = Employee(1, "Noha", 10000)
print(Employee.cal_net_salary(emp1.salary))
print(emp1.cal_net_salary(54545))
```
